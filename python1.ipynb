```python
n= input("Enter a number: ")

if n== n[::-1]:
    print(n, "is a Palindrome")
else:
    print(n, "is not a Palindrome")
```

```python
n=int(input("Enter a no.: "))
r=0
while n>0:
  a=n%10
  r=r*10+a
  n=n//10
print(r)
```

```python
n=int(input("enter a number"))
sum=0
while n>0:
  sum+=n%10
  n=n//10
print(sum)
```

```python
n=int(input("Enter a number: "))
s=0
while n>0:
  a=n%10
  if a>s:
    s=a
  n=n//10
print("Largest digit: ",s)
```
